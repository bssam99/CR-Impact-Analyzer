## Notebook 1 — Setup & Configuration

###  Step 1 — Configure your settings here

```python
import os

BASE_DIR         = os.getcwd()                          # folder where this notebook lives
PDF_DIR          = os.path.join(BASE_DIR, "pdfs")       # drop your PDFs here
VECTOR_STORE_DIR = os.path.join(BASE_DIR, "vector_store") # ChromaDB will be saved here

COLLECTION_NAME = "system_docs"

CHUNK_SIZE    = 1000   # characters per chunk
CHUNK_OVERLAP = 200    # overlap between consecutive chunks

EMBEDDING_MODEL  = "all-MiniLM-L6-v2"  # fast & accurate, runs fully offline
EMBED_BATCH_SIZE = 64

OLLAMA_BASE_URL = "http://localhost:11434"
LLM_MODEL       = "llama3"   # change to: mistral / phi3 / gemma:7b

TOP_K = 6   # number of chunks to retrieve per query

print(" Configuration loaded")
print(f"   PDF folder       : {PDF_DIR}")
print(f"   Vector store     : {VECTOR_STORE_DIR}")
print(f"   LLM model        : {LLM_MODEL}")
print(f"   Embedding model  : {EMBEDDING_MODEL}")
```

###  Step 2 — Create required folders

```python
os.makedirs(PDF_DIR, exist_ok=True)
os.makedirs(VECTOR_STORE_DIR, exist_ok=True)

print(" Folders ready")
print(f"  Drop your PDF files into:  {PDF_DIR}")
```

###  Step 3 — Check all imports are working

```python
print("Checking imports...\n")

import pymupdf as fitz
print(f"   PyMuPDF           {fitz.__version__}")

import langchain
print(f"   LangChain         {langchain.__version__}")

from langchain_text_splitters import RecursiveCharacterTextSplitter
print(f"   Text splitter     OK")

import sentence_transformers
print(f"   SentenceTransformers  {sentence_transformers.__version__}")

import chromadb
print(f"   ChromaDB          {chromadb.__version__}")

import requests
print(f"   Requests          {requests.__version__}")

print("\n All imports OK — environment is ready!")
```

### Step 4 — Check Ollama is running

```python
import requests

try:
    r = requests.get(f"{OLLAMA_BASE_URL}/api/tags", timeout=5)
    models = [m["name"] for m in r.json().get("models", [])]
    print(f" Ollama is running at {OLLAMA_BASE_URL}")
    print(f"   Available models: {models}")

    if any(LLM_MODEL in m for m in models):
        print(f"\n Model '{LLM_MODEL}' is ready to use!")
    else:
        print(f"\n  Model '{LLM_MODEL}' not found.")
        print(f"   Run in a terminal:  ollama pull {LLM_MODEL}")

except Exception as e:
    print(" Ollama is NOT running.")
    print("    Open a terminal and run:  ollama serve")
```

### 📄 Step 5 — Check PDFs in your folder
